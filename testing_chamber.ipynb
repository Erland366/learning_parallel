```python
import torch
from torch import nn
from einops import rearrange
from torch.nn import functional as F




class Attention(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.q_proj = nn.Linear(hidden_dim, hidden_dim)
        self.k_proj = nn.Linear(hidden_dim, hidden_dim)
        self.v_proj = nn.Linear(hidden_dim, hidden_dim)
        self.o_proj = nn.Linear(hidden_dim, hidden_dim)

    def forward(self, x):
        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)
        attention = F.softmax(q @ k.T, dim=-1) / torch.sqrt(self.hidden_dim) # B, S, H, S
        o = self.o_proj(attention @ v)
        return o

attention = Attention(128).to("cuda")
example_input = torch.randn((2, 8, 4, 128), dtype=torch.bfloat16, device="cuda")
output = attention(example_input)
print(output)

```
