```python
import torch

dp_size = 2
pp_size = 2
tp_size = 2
global_rank = 3

grid = torch.arange(dp_size * pp_size * tp_size).view(dp_size, pp_size, tp_size)
dp_rank, pp_rank, tp_rank = (grid == global_rank).nonzero().flatten().tolist()
print(dp_rank, pp_rank, tp_rank)
```
